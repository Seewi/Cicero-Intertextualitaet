```python
import re
import pandas as pd
```

```python
briefe = pd.read_csv("../data/cicero_letters.csv")
```

```python
matcher = re.compile(r"\b(?P<author>Homer\w*|Menand\w*|Strattis\w*|Epicharm\w*|E[uv]ripid\w*|Rh?inth?on\w*|Enni\w*|Sophocl\w*|Lucili\w*|Terent\w*|Plato\w*|Aeschyl\w*|Naevi\w*|Aristophan\w*|Archiloch\w*|Stati\w*|Trabea|Isocrat\w*|Th[uv]cydid\w*|Carnead\w*|Caecili|Phocyl\w*|Diphil\w*|Stesichor\w*|Herodot\w*|Pindar\w*|Acci[uv]\w*|Epic[uv]r\w*|Hesiod\w*|T[uv]rpili\w*|Pac[uv]vi\w*|(Piso\w*.*?Fr[uv]gi\w*|Fr[uv]gi\w*.*?Piso\w*)|Pomponi\w*|Atili\w*|Afran\w*|Pla[uv]t\w*|Cat[uv]ll\w*|Livi\w*.*?Andron\w*|Polyb\w*|(P[uv]bli[l1i]i?us\w*\s+Syr\w*|Syr\w*\s+P[uv]bli[l1i]i?i?us\w*)|T[uv]rp\w*|L[uv]creti\w*|(Cornel[i1l]i\w*\s+Nepos\w*|Nepos\w*\s+Cornel[i1l]i\w*)|(Corneli\w*\s+Nepo\w*|Nep\w*\s+Corneli\w*)|Sall[uv]st\w*|Fabi\w*\s+Pict\w*)\b")
```

```python
df_matches=briefe["text"].str.extractall(matcher)
```

```python
matches_rejoined = df_matches.join(briefe, how='left')
matches_rejoined=matches_rejoined.drop(columns = [1])


```

```python
matches_rejoined.to_csv("../data/matches.csv",index=False)
```
