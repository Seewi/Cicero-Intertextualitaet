```python
# Problematic Patterns:

# Afranius hat keine direkten Matches, nur 1x zitiert, aber ohne Namen, Afran\w*

probpattern = r"""(?<!\w)(?P<author>
Caecilianam|
dico\sCaecilium
)(?!\w)"""
```

```python
matcher = re.compile(pattern, re.VERBOSE)
```

```python
prob_matcher = re.compile(probpattern, re.VERBOSE)
```

```python
df_matches = briefe["text"].str.extractall(matcher)
```

```python
df_prob = briefe["text"].str.extractall(prob_matcher)
```

```python
matches_rejoined = df_matches.join(briefe, how='left')
```
