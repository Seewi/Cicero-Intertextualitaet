```python
import re
import pandas as pd
```

```python
briefe = pd.read_csv("../data/cicero_letters.csv")
```

```python
matcher = re.compile(r"\b(?P<author>Homer\w*|Menand\w*|Strattis\w*|Epicharm\w*|Euripid\w*|Rh?inth?on\w*|Enni\w*|Sophocl\w*|Lucili\w*|Terent\w*|Plato\w*|Aeschyl\w*|Naevi\w*|Aristophan\w*|Archiloch\w*|Stati\w*|Trabea|Isocrat\w*|Thucydid\w*|Carnead\w*|Caecili|Phocyl\w*|Diphil\w*|Stesichor\w*|Herodot\w*|Pindar\w*|Acciu\w*|Epicur\w*|Hesiod\w*|Turpili\w*|Pacuvi\w*|(Piso\w*.*?Frugi\w*|Frugi\w*.*?Piso\w*)|Pomponi\w*|Atili\w*|Afran\w*|Plaut\w*|Catull\w*|Livi\w*.*?Andron\w*|Polyb\w*)\b")
```

```python
df_matches=briefe["text"].str.extractall(matcher)
```
